```python
# step 1: given data
customers = [
"Ahmed", "Sara", "Omar", "Mona",
"Ali", "Nour", "Youssef", "Hana",
"Khaled", "Mariam", "Hassan", "Laila"
] 
scores = [
91, 45, 78, 66, 88, 30,
72, 95, 51, 60, 84, 39
]

#step 2,3: classify customers , calculate statistics
risk_levels=[]
total=0
cntH=0
cntM=0
cntL=0
for c,s in zip(customers,scores):
    if 85<=s<=100:
        risk_levels.append("High")
        cntH+=1
    elif 60<=s<85:
        risk_levels.append("Medium")
        cntM+=1
    else:
        risk_levels.append("Low")
        cntL+=1
    total+=s

avg_risk_scores=total/len(scores)

#step 4: best & worst customer
mx=0
mn=101
mx_i=-1
mn_i=-1
for i in range(len(scores)):
    if scores[i]>mx:
        mx=scores[i]
        mx_i=i
    if scores[i]<mn:
        mn=scores[i]
        mn_i=i


#step 6: menu
while(True):
    print()
    print("========== MENU ==========")
    print("1. Show All Customers")
    print("2. Search for a Customer")
    print("0. Exit")
    print()
    choice=int(input("Enter your choice: "))
    if choice==0:
        print("Goodbye!")
        break

    elif choice ==2:
        name=input("Enter customer name: ")
        if name in customers:
            indx=customers.index(name)
            print(f"Name: {name}")
            print(f"Score: {scores[indx]}")
            print(f"Risk: {risk_levels[indx]}")
        else:
            print("Customer not found!")

    elif choice==1:
        #step 5: report
        print("========== CUSTOMER RISK REPORT ==========")
        print("Name     Score     Risk")
        print("------------------------------------------")
        for c,s,r in zip(customers,scores,risk_levels):
         print(c,"     ",s,"    ",r)
        print("=============== SUMMARY ================")
        print(f"High Risk: {cntH}")
        print(f"Medium Risk: {cntM}")
        print(f"Low Risk: {cntL}")
        print(f"Average Score: {avg_risk_scores:.2f}")
        print(f"Best Customer: {customers[mn_i]} ({scores[mn_i]})")
        print(f"Worst Customer: {customers[mx_i]} ({scores[mx_i]})")
    
    else: 
        print("Invalid Choice")
```
