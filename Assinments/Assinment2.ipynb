Program 1

```python
members=int(input("Enter the Number of team members: "))
months=int(input("Enter the Number of months the project will take: "))
salary=float(input("Enter the amount of monthly salary: "))
cloud=float(input("Enter Total cloud cost for the whole project: "))

salary_cost=members*months*salary
subtotal=salary_cost+cloud
VAT=subtotal*0.14
total=subtotal+VAT

print(f"The Salary cost= {salary_cost:,.2f}, The Subtotal= {subtotal:,.2f}, VAT= {VAT:,.2f}, The Total= {total:,.2f}")
```

```python
message=input("Enter your message: ")
hours=float(input("Enter How many hours the ticket has been open: "))

if "payment" in message:
    print("Team:Billing")
elif "delivery" in message:
    print("Team:Logistics")
elif "error" in message:
    print("Team:Tech")
else:
    print("Team:General")

if hours>=24:
    print("Priority:HIGH")
else:
    print("Priority:LOW")
```
